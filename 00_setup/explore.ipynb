## 1. Running a cell

A notebook is a list of cells. Click the cell below and press **Shift+Enter**. The output appears
underneath it.

````python
print("Hello. The tools work.")
````

That is the whole mechanic. One warning, because it costs beginners hours:

**Cells remember.** Everything you create in one cell is still there in the next one. If you go
back up and change a cell, the value below does *not* update until you run it again. When output
stops making sense, use **Kernel → Restart Kernel and Run All Cells** and read from the top.

````python
reading = 21.7
print(reading)
````

````python
reading = reading + 1
print(reading)
````

Run that last cell three times in a row and watch the number climb. Nothing is broken —
the cell simply ran three times. That is the trap; now you know it.

## 3. Reading an error message

Errors are not a sign that you did something stupid. They are the machine telling you precisely
what it could not do. Learning to read them is the single most useful skill in this course.

The next cell is broken on purpose. Run it.

````python
reading = "21.7"
limit = 25

print(limit - reading)
````

**Read it bottom-up.**

1. **The last line** says *what* happened:
   `TypeError: unsupported operand type(s) for -: 'int' and 'str'`
   You tried to subtract text from a number.
2. **The lines above** say *where* — file and line number. The innermost place is at the bottom.
3. **Only then** look at the code.

The fix is one conversion:

````python
reading = "21.7"
limit = 25

print(limit - float(reading))
````

### About that type hint

The function is written like this:

```python
def to_fahrenheit(celsius: float) -> float:
```

`celsius: float` says the function expects a decimal number; `-> float` says it hands one back.

**Python does not enforce this while running.** It is a promise to whoever reads the code, and
`mypy` is the tool that checks whether you keep it. Run the cell below to see the difference
between a promise kept and a promise broken.

````python
def to_fahrenheit(celsius: float) -> float:
    return "warm"  # a lie: this is text, not a number


print(to_fahrenheit(21.5))  # Python runs it anyway
````

It printed `warm`. Python did not care. But `uv run mypy` would say:

```
error: Incompatible return value type (got "str", expected "float")
```

That is the point of type checking: it finds the lie **before** the program runs, in a place
where a real program would have crashed three steps later with a much more confusing message.

You will meet type hints properly in module 04. For now it is enough to know the command and
what it is for.

## 5. How a module is laid out

Every module folder looks the same:

| | |
