## 2. Cells remember

A notebook keeps a Python process running behind it — the *kernel*. Everything a
cell creates stays in memory until you restart, which is the point: you build
something expensive once and then ask it fifty cheap questions.

The cost is that the notebook has a state that does not match the order of the
cells on screen. Run the next two, then run the second one three more times.

````python
reading = 21.7
print(reading)
````

````python
reading = reading + 1
print(reading)
````

The number climbs, because the cell ran four times. Nothing is wrong; the screen
shows a *program* and the kernel holds a *state*, and the two drift apart as soon
as you edit out of order.

When output stops making sense: **Kernel → Restart Kernel and Run All Cells**,
then read from the top. That is also how you check that a notebook you are about
to hand over actually runs.

## 4. Reading a traceback

The next cell is broken on purpose. Run it.

````python
reading = "21.7"
limit = 25

print(limit - reading)
````

Note the **direction**: Python prints the innermost frame **last** — that is what
"most recent call last" means. A Java stack trace prints it first. Read a Python
traceback bottom-up, or you will land on the caller instead of on the crash.

The exception type and message are the final line; the frames above it are the
call chain. Here: `TypeError: unsupported operand type(s) for -: 'int' and 'str'`.

The fix is one conversion:

````python
reading = "21.7"
limit = 25

print(limit - float(reading))
````

## 6. What a type hint is for

```python
def to_fahrenheit(celsius: float) -> float:
```

Unlike C or Java, Python does not enforce that. The annotation is metadata; the
interpreter ignores it entirely, and nothing stops the function from returning a
string. `mypy` is the separate tool that checks whether you kept the promise.

````python
def to_fahrenheit(celsius: float) -> float:
    return "warm"  # a lie: this is text, not a number


print(to_fahrenheit(21.5))  # runs, prints "warm", no complaint
````

`uv run mypy` on the same code says:

```
error: Incompatible return value type (got "str", expected "float")
```

That is the trade. A compiler would have refused the function and you would never
think about it again; here you get to prototype without annotations, and you get
nothing back unless you run the checker. Which is why the course runs it in CI.
