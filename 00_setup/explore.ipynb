## 2. Cells remember

This is the one thing about notebooks worth knowing before anything else,
because not knowing it costs beginners hours.

**A notebook keeps a live Python process running behind it** — the *kernel*.
Everything you create in one cell is still there in the next one, and stays
there until you restart. That is what makes a notebook useful for exploring:
you can build something expensive once and then ask it fifty cheap questions.

It also means the notebook has a memory that does not match what you see on the
screen. Run the next two cells in order.

````python
reading = 21.7
print(reading)
````

````python
reading = reading + 1
print(reading)
````

Now run that second cell **three more times** and watch the number climb.
Nothing is broken: the cell simply ran four times, and each run added one.

The screen shows a *program*; the kernel holds a *state*. When output stops
making sense, the fix is always the same: **Kernel → Restart Kernel and Run All
Cells**, then read from the top.

## 3. Predicting, with `assert`

**How the predictions work.** Some cells look like this:

```python
assert 7 // 2 == ...
```

`...` is a real value in Python (it is called `Ellipsis`), and it is never equal
to a number — so the cell fails until you replace it with your answer. Run it
once before you edit it, to see what a failed prediction looks like. When every
`...` is replaced correctly the cell prints nothing at all. **Silence means
right.**

Guess rather than skip. A wrong guess marks the exact spot where your idea of
the machine differs from the machine, and that is the only thing here worth
spending time on.

`assert` is not a notebook feature — it is ordinary Python, and you will meet it
again in every test in this course. It says: *this must be true; if it is not,
stop here and say so.*

Try the cell below. Run it once unchanged to see the failure, then replace the
`...` and run it again.

````python
assert 3 + 4 == ...
````

## 4. Reading an error message

Errors are not a sign that you did something stupid. They are the machine
telling you precisely what it could not do, and reading them is the single most
useful skill in this course.

The next cell is broken on purpose. Run it.

````python
reading = "21.7"
limit = 25

print(limit - reading)
````

**Read it bottom-up.**

1. **The last line** says *what* happened:
   `TypeError: unsupported operand type(s) for -: 'int' and 'str'`
   You tried to subtract text from a number.
2. **The lines above** say *where* — file and line number. The innermost place
   is at the bottom, which is what "most recent call last" means.
3. **Only then** look at the code.

The fix is one conversion:

````python
reading = "21.7"
limit = 25

print(limit - float(reading))
````

## 6. What a type hint is for

The function you are about to complete is written like this:

```python
def to_fahrenheit(celsius: float) -> float:
```

`celsius: float` says the function expects a decimal number; `-> float` says it
hands one back.

**Python does not enforce this while running.** It is a promise to whoever reads
the code, and `mypy` is the tool that checks whether you keep it. Run the cell
below to see a promise being broken with no complaint at all.

````python
def to_fahrenheit(celsius: float) -> float:
    return "warm"  # a lie: this is text, not a number


print(to_fahrenheit(21.5))  # Python runs it anyway
````

It printed `warm`. Python did not care. `uv run mypy` would say:

```
error: Incompatible return value type (got "str", expected "float")
```

That is what a type checker buys you: the lie is found **before** the program
runs, at the place where it was told — instead of three steps later, in a
crash whose message points somewhere else entirely.

> **If you know Java or C:** those languages refuse to compile that function at
> all, and you never think about it again. Python's type hints are optional and
> checked by a separate tool, which is a genuine trade: you can prototype
> without them, and you get nothing for free unless you run `mypy`.

You will meet type hints properly in module 04. For now it is enough to know
the command and what it is for.
