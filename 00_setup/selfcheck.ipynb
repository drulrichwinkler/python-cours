Set to `True` what you can do **right now, without looking it up**, then run the
second cell. It tells you whether this module has anything for you.

Nothing is recorded and nothing is graded. An honest `False` here saves more time
than an optimistic `True`.

```python
can = {
    "run `uv sync` and say what it built, and where": False,
    "say what `uv run pytest` does that a bare `pytest` does not": False,
    "run the tests of one module, and of one single exercise": False,
    "read a Python traceback and name the line that raised": False,
    "say what a type annotation does at runtime, and what `mypy` adds on top": False,
    "say why a notebook can print a value that no longer matches the code above it": False,
}
```

```python
open_items = [q for q, yes in can.items() if not yes]

print(f"{len(can) - len(open_items)} of {len(can)}\n")

if not open_items:
    print("Skip this module. Skim explore.ipynb if you want to confirm the details.")
elif len(open_items) <= 2:
    print("Mostly there. Read explore.ipynb, then do the exercises for what is left:\n")
else:
    print("Worth working through. These are what the module is for:\n")

for q in open_items:
    print(" -", q)
```
