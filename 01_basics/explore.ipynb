## 6. f-strings

`temperature = 23.456`. What does `f"{temperature:.1f}"` print?

**Type your prediction into the next cell and run it.**
Only then run the cell below that.

````python
prediction = "..."  # <- fill this in

check(
    prediction,
    "e378bdd3a01cdff854a7130fe2adb8e001441fd42c7718e49c76f9b1cf51aa27",
    "One decimal place -- and it rounds, it does not truncate.",
)
````

````python
temperature = 23.456
print(f"It is {temperature} degrees.")
print(f"It is {temperature:.1f} degrees.")
print(f"{temperature=}")
````

> The last line is the most useful trick in this module: the `=` **inside** the braces prints
> the name *and* the value. When you want to know what is in a variable, that is the shortest
> question you can ask.

## 8. The most famous surprise

What is `0.1 + 0.2`? Write it out **exactly**, digit by digit.

**Type your prediction into the next cell and run it.**
Only then run the cell below that.

````python
prediction = "..."  # <- fill this in

check(
    prediction,
    "06bad31060c1212ae832de4c031f7b31e3b48aed57858294478cb19450cf34ca",
    "It is not 0.3. Count the zeros.",
)
````

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

`0.1` cannot be represented exactly in binary — the same way ⅓ cannot be represented exactly
in decimal. **Every** language with floating point numbers behaves this way; they all follow the
same IEEE 754 standard.

→ Never compare floats with `==`. Never store money as a float.
