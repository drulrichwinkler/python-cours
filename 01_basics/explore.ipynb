## 1. Printing

`print()` writes its arguments to the screen. Several arguments are separated
by a space, and the line ends with a newline — unless you say otherwise.

````python
print("Sensor", "Temperature", "OK")
print("Sensor", "Temperature", "OK", sep=" | ")
print("no newline after this", end="")
print(" <- so this continues the line")
````

`sep=` is what goes *between* the arguments, `end=` what goes *after the last*.
Both are ordinary arguments with a default you are overriding.

## 2. Names and values

A variable in Python is a **name pointing at a value**. There is no declaration
and no type to announce: you write a name, `=`, and a value, and from then on
the name refers to it.

The `=` is not the `=` of mathematics. It is not a claim that two things are
equal — it is an instruction: *make this name refer to that value.* Which is
why `a = a + 1` is a sensible thing to write and `x = y` says nothing about
what `y` does later.

````python
name = "TH-04"
reading = 21.7
is_active = True

print(name, reading, is_active)
````

## 3. Whole numbers, decimals, and two kinds of division

`int` is a whole number, `float` a decimal one. Python has **two** division
operators, and the difference catches almost everybody once.

- `/` is **true division** and always produces a `float`, even when it comes out
  even: `6 / 3` is `2.0`, not `2`.
- `//` is **floor division**: divide, then throw the fraction away.
- `%` is the **remainder** left over.

The last two fit together: `7 // 2` is `3`, `7 % 2` is `1`, and `3 * 2 + 1 = 7`.
`%` is the most underrated operator in the language — `n % 2 == 0` asks whether
`n` is even, and `seconds % 60` gives the seconds within the minute.

> **If you know C or Java:** there, `7 / 2` between two integers *is* integer
> division and gives `3`. Python's `/` never does that. `//` is the operator
> that behaves the way you are used to.

````python
assert 6 / 3 == ...
assert 7 // 2 == ...
assert 7 % 2 == ...
````

Another difference worth knowing early: **`int` in Python has no upper limit.**
It grows as large as your memory allows.

````python
print(2**100)
````

> **If you know C or Java:** an `int` there is 32 bits and silently wraps around
> when it overflows; `long` buys you 64 and then does the same. Python simply
> does not have that failure mode. A great deal of defensive code you may have
> written elsewhere is unnecessary here.

## 5. The type belongs to the value

You never declare a type in Python. The value carries it, and `type()` will tell
you which one.

````python
print(type(42))
print(type(42.0))
print(type("42"))
print(type(True))
print(type(None))
````

Note the third line: `"42"` in quotes is **text that looks like a number**, and
that distinction causes more beginner bugs than any other single thing.

`None` is its own type. It means *there is no value here* — not zero, not empty
text, not false. It is what a function hands back when it has nothing to say.

**Predict:** `is` compares identity, and every type is itself, so this works.

````python
assert type(21.7) is ...
assert type("21.7") is ...
````

> **If you know Java or C:** there the *variable* has a type, fixed when you
> declare it, and the compiler stops you from putting the wrong thing in. In
> Python the *value* has the type and a name can point at anything, so nothing
> stops you from writing `x = 5` and then `x = "five"`. That is freedom you pay
> for: mistakes of that kind surface when the program runs, which is exactly
> why `mypy` and type hints exist.

## 6. Converting between types

`int()`, `float()` and `str()` build a new value of that type from another one.
Two details are worth memorising:

- **`int()` truncates, it does not round.** `int(3.9)` is `3`. If you want
  rounding, that is `round()`.
- **`int()` on text only accepts a whole number.** `int("42")` works,
  `int("3.9")` raises `ValueError` — because it has to *interpret* the text, and
  `"3.9"` does not spell a whole number. The way round is `int(float("3.9"))`.

````python
assert int(3.9) == ...
assert round(3.9) == ...
assert float("3.14") == ...
````

## 7. What counts as true

Every value in Python can be used as a condition, and `bool()` tells you what it
counts as. The rule is short: **empty is false, everything else is true.**

Zero, the empty string, the empty list — false. Any other number, any text with
at least one character in it — true.

````python
print(bool(0), bool(1), bool(-1))
print(bool(""), bool(" "), bool([]))
````

That last rule has one famous consequence. **Predict it:**

````python
assert bool("False") is ...
````

`"False"` is text with five characters in it. `bool()` does not ask *what* the
text says, only *whether there is any* — so it is true. The word inside is a
coincidence.

> **If you know C:** this will feel familiar, since C also treats `0` as false
> and everything else as true. **If you know Java:** it will not — there a
> condition must actually be a `boolean`, and `if ("False")` does not compile.

## 8. Building strings with f-strings

Putting values into text is so common that Python has syntax for it. An `f`
before the quote turns `{...}` into "evaluate this and put the result here".

````python
tag = "TH-04"
reading = 22.83333

print(f"{tag} reports {reading} degrees")
print(f"{tag} reports {reading:.1f} degrees")
print(f"next reading is number {3 + 1}")
````

`:.1f` is a *format spec*: show this as a decimal number with one digit after
the point. It changes only the output — `reading` still holds all its digits.

And one small trick worth keeping:

````python
reading = 22.83333
print(f"{reading=}")
````

The `=` **inside** the braces prints the name and the value. When you want to
know what is in a variable, that is the shortest question you can ask.

## 9. Why decimals are not exact

Now the one that surprises everyone. **Predict the value of `0.1 + 0.2`** —
write out every digit.

````python
assert 0.1 + 0.2 == ...
````

The answer is `0.30000000000000004`, and it is not a bug.

`0.1` cannot be written exactly in binary, in the same way `1/3` cannot be
written exactly in decimal — `0.3333...` never terminates. The computer stores
the closest value it can, and adding two of those makes the gap visible.

> **This is not a Python quirk.** C, Java, JavaScript and Go all do the same,
> because they all follow the same IEEE 754 standard for floating point. Try it
> in any of them.

Two rules follow, and they are worth taking seriously:

- **Never compare floats with `==`.** Compare the difference against a small
  tolerance instead. The tests in this course use `pytest.approx` for exactly
  this reason.
- **Never store money as a float.** Use whole cents as an `int`, or
  `decimal.Decimal`.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
````
