## 1. Printing

`print()` takes any number of arguments, separates them with a space and ends the
line. Both of those are parameters with defaults you can override.

```python
print("Sensor", "Temperature", "OK")
print("Sensor", "Temperature", "OK", sep=" | ")
print("no newline after this", end="")
print(" <- so this continues the line")
```

## 2. Names

value, and the value carries the type.

```python
name = "TH-04"
reading = 21.7
is_active = True

print(name, reading, is_active)
```

## 3. Two kinds of division

In C and Java, `7 / 2` between two ints is `3`. In Python it is `3.5`: `/` is
always true division and always produces a `float`, even when it comes out even —
`6 / 3` is `2.0`, not `2`.

`//` is the integer one, and `%` the remainder that goes with it: `7 // 2` is `3`,
`7 % 2` is `1`, and `3 * 2 + 1 = 7`.

**`//` floors, it does not truncate.** For positive operands it matches C, but
`-7 // 2` is `-4` here where C and Java give `-3`. Python rounds towards negative
infinity; C rounds towards zero.

```python
assert 6 / 3 == ...
assert 7 // 2 == ...
assert -7 // 2 == ...
assert 7 % 2 == ...
```

## 4. Integers do not overflow

There is no 32-bit `int` and no 64-bit `long`. Python integers grow to whatever
your memory allows, so the wrap-around you guard against in C simply is not a
failure mode here.

```python
print(2**100)
print(len(str(2**10000)), "digits")
```

## 5. Text

Single and double quotes are interchangeable — pick the one that saves you an
escape. Triple quotes span lines.

```python
print("Python" == "Python")
print('She said: "yes"')
print("""first line
second line""")
```

## 6. The type belongs to the value, not to the name

In C and Java the *variable* has a type, fixed at declaration, and the compiler
stops you putting the wrong thing in it. In Python the *value* has the type, and
a name can be rebound to anything at all.

```python
x = 5
print(x, type(x))

x = "five"
print(x, type(x))  # same name, different type, no complaint
```

Nothing there is an error. That is the freedom you are paying for when a
`TypeError` shows up at runtime instead of at compile time — and the reason type
hints and `mypy` exist at all.

`type()` reports what a value actually is:

```python
print(type(42))
print(type(42.0))
print(type("42"))
print(type(True))
print(type(None))
```

`"42"` in quotes is text that looks like a number, and that distinction causes
more bugs in this course than anything else — exercise 03 is exactly that crash.

`None` is its own type, and it is Python's `null`: what a function returns when it
has nothing to return. Unlike `null` it is a real object, so `None.__class__`
works and there is no null pointer to dereference.

**Predict.** Every type is itself, so `is` is the right comparison:

```python
assert type(21.7) is ...
assert type("21.7") is ...
```

## 7. Converting

`int()`, `float()` and `str()` build a new value from another one. Two details:

- **`int()` truncates towards zero, it does not round.** `int(3.9)` is `3`.
  Rounding is `round()`, and it rounds half to even — `round(0.5)` is `0`.
- **`int()` on text accepts only a whole number.** `int("42")` works,
  `int("3.9")` raises `ValueError`. The way round is `int(float("3.9"))`.

```python
assert int(3.9) == ...
assert round(3.9) == ...
assert float("3.14") == ...
```

## 8. Truthiness

Every value can stand in a condition. The rule: **empty is false, everything else
is true** — zero, the empty string, the empty list, `None` are false; anything
else is true.

This is C's `0`/non-zero rule generalised to every type. Java's stricter rule
does not apply: `if (someString)` compiles here and means "is it non-empty".

```python
print(bool(0), bool(1), bool(-1))
print(bool(""), bool(" "), bool([]), bool(None))
```

Which produces one famous consequence. **Predict it:**

```python
assert bool("False") is ...
```

`"False"` is five characters. `bool()` asks whether there is any text, not what
it says.

## 9. f-strings

An `f` before the quote makes `{...}` an expression slot. This is Python's
equivalent of `String.format` or `printf`, with the arguments written where they
appear.

```python
tag = "TH-04"
reading = 22.83333

print(f"{tag} reports {reading} degrees")
print(f"{tag} reports {reading:.1f} degrees")
print(f"next reading is number {3 + 1}")
```

`:.1f` is a format spec — one digit after the point, output only; `reading` keeps
all its digits.

And `=` **inside** the braces prints the name with the value, which is the
shortest debug statement in the language:

```python
reading = 22.83333
print(f"{reading=}")
```

## 10. Floating point

**Predict `0.1 + 0.2`** — every digit.

```python
assert 0.1 + 0.2 == ...
```

`0.1` has no exact binary representation, the same way `1/3` has no exact decimal
one. The stored value is slightly off, and adding two of them makes the gap
visible at the seventeenth digit.

This is IEEE 754, so C, Java, JavaScript and Go all produce the same thing —
nothing here is Python-specific except that Python prints enough digits to show
it. Java's `System.out.println(0.1 + 0.2)` prints `0.30000000000000004` too.

Consequences worth taking seriously:

- **Never compare floats with `==`.** Compare against a tolerance. The tests in
  this course use `pytest.approx` for exactly that.
- **Never store money as a float.** Integer cents, or `decimal.Decimal`.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
```
