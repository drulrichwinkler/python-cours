## 6. f-Strings

`temperatur = 23.456`. Was gibt `f"{temperatur:.1f}"` aus?

**Tragen Sie Ihre Vorhersage in die nächste Zelle ein und führen Sie sie aus.**
Erst danach die Zelle darunter.

````python
vorhersage = "..."  # <- hier eintragen

pruefe(
    vorhersage,
    "e378bdd3a01cdff854a7130fe2adb8e001441fd42c7718e49c76f9b1cf51aa27",
    "Eine Nachkommastelle -- und es wird gerundet, nicht abgeschnitten.",
)
````

````python
temperatur = 23.456
print(f"Es sind {temperatur} Grad.")
print(f"Es sind {temperatur:.1f} Grad.")
print(f"{temperatur=}")
````

> Die letzte Zeile ist der nützlichste Trick des Moduls: Das `=` **in** der Klammer druckt
> Name *und* Wert. Wenn Sie wissen wollen, was in einer Variablen steckt, ist das die kürzeste
> Frage, die Sie stellen können.

## 8. Die berühmteste Überraschung

Was ergibt `0.1 + 0.2`? Schreiben Sie es **genau** hin, Ziffer für Ziffer.

**Tragen Sie Ihre Vorhersage in die nächste Zelle ein und führen Sie sie aus.**
Erst danach die Zelle darunter.

````python
vorhersage = "..."  # <- hier eintragen

pruefe(
    vorhersage,
    "06bad31060c1212ae832de4c031f7b31e3b48aed57858294478cb19450cf34ca",
    "Es ist nicht 0.3. Zaehlen Sie die Nullen.",
)
````

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

`0.1` lässt sich im Zweiersystem nicht exakt darstellen — so wie ⅓ im Dezimalsystem nicht
exakt darstellbar ist. Das hat **jede** Sprache mit Fließkommazahlen.

→ Fließkommazahlen nie mit `==` vergleichen. Geldbeträge gar nicht erst als `float` speichern.
