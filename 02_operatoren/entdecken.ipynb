## 6. Ein Statusbyte auslesen

Ein Sensor meldet seinen Zustand als ein Byte:

```
0 0 0 0 0 1 1 0
              └─ Bit 0: Messung bereit
            └─── Bit 1: Grenzwert überschritten
          └───── Bit 2: Sensorfehler
        └─────── Bit 3: Kalibrierung nötig
```

`status = 0b0110`, `GRENZWERT = 0b0010`. Was ergibt `status & GRENZWERT`?

**Tragen Sie Ihre Vorhersage in die nächste Zelle ein und führen Sie sie aus.**
Erst danach die Zelle darunter.

````python
vorhersage = "..."  # <- hier eintragen

pruefe(
    vorhersage,
    "d4735e3a265e16eee03f59718b9b5d03019c07d8b6c51f90da3a666eec13ab35",
    "Das & blendet alles aus ausser dem einen Bit. Uebrig bleibt eine Zahl, nicht True.",
)
````

````python
BEREIT, GRENZWERT, FEHLER, KALIBRIERUNG = 0b0001, 0b0010, 0b0100, 0b1000
status = 0b0110

print(status & GRENZWERT)
print(status & BEREIT)
print(bool(status & GRENZWERT))
print(bin(status | BEREIT))
print(bin(status & ~FEHLER))
````

**Wo Ihnen das begegnet:** GPIO-Pins am Mikrocontroller, Statusregister in Datenblättern,
Berechtigungen im Dateisystem (`chmod 755`), Flags in Netzwerkprotokollen. Überall dort, wo
viele Ja/Nein-Angaben in eine Zahl passen müssen.

## 7. Gleich, oder dasselbe?

```python
a = [1, 2]
b = [1, 2]
```
Was ergibt `a is b`?

**Tragen Sie Ihre Vorhersage in die nächste Zelle ein und führen Sie sie aus.**
Erst danach die Zelle darunter.

````python
vorhersage = "..."  # <- hier eintragen

pruefe(
    vorhersage,
    "fcbcf165908dd18a9e49f7ff27810176db8e9f63b4352213741664245224f8aa",
    "== fragt nach dem Wert, is nach der Identitaet im Speicher.",
)
````

````python
a = [1, 2]
b = [1, 2]
print(a == b)
print(a is b)
````

| | Frage |
|---|---|
| `==` | Haben beide **denselben Wert**? |
| `is` | Sind beide **dasselbe Objekt** im Speicher? |

> `is` beantwortet eine Frage, die Sie fast nie stellen wollen. Für Werte nehmen Sie `==`.
> `is` benutzen Sie für genau einen Fall: `if ergebnis is None`.

## 8. Ketten sind keine Klammern

`3 > 2 > 1` ergibt `True`. Was ergibt `(3 > 2) > 1`?

**Tragen Sie Ihre Vorhersage in die nächste Zelle ein und führen Sie sie aus.**
Erst danach die Zelle darunter.

````python
vorhersage = "..."  # <- hier eintragen

pruefe(
    vorhersage,
    "fcbcf165908dd18a9e49f7ff27810176db8e9f63b4352213741664245224f8aa",
    "In der Klammer entsteht True -- und True verhaelt sich in einer Rechnung wie 1.",
)
````

````python
print(3 > 2 > 1)
print((3 > 2) > 1)
print(True + True)
````

`1 < x < 10` ist die pythonische Bereichsprüfung und meint wirklich beide Vergleiche.
Mit Klammern wird daraus etwas anderes: `True > 1` heißt `1 > 1` und damit `False`.
